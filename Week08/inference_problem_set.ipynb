```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
import seaborn as sns 

from scipy.stats import norm

df = pd.read_csv('./data/ames_prices.csv')
df.head()
```

4. For `./data/ames_prices.csv`, define an indicator equal to 1 when `Central.Air == 'Y'`. Estimate the proportion of homes with central air and test $H_0:p=.90$. Treat a proportion as the mean of a binary variable and complete the full inference routine.

```python
df["air_indicator"] = (df["Central.Air"] == "Y").astype(int)
print(f"Proportion no central air: {df['air_indicator'].value_counts()[0]/df['air_indicator'].value_counts().sum()}")
print(f"Proportion central air: {df['air_indicator'].value_counts()[1]/df['air_indicator'].value_counts().sum()}")
p_hat = df["air_indicator"].mean()
n = len(df)

print("Sample size:", n)
```

```python
rng = np.random.default_rng(42)
R = 2_000

x = df["air_indicator"].to_numpy()


results = []
for t in range(R):
    bootstrap_sample = rng.choice(x, size=n, replace=True)
    p_hat = bootstrap_sample.mean()
    results.append(p_hat)
results = np.array(results)
```

```python
se = results.std(ddof=1)
p_hat = df["air_indicator"].mean()
se
```

```python
for z_star in [1.645, 1.96, 2.576]:
    lower = p_hat - z_star * se
    upper = p_hat + z_star * se
    print(f"CI ({z_star}): ({lower:.4f}, {upper:.4f})")
```

```python
z = (p_hat - 0.90) / se
p_value = 2 * norm.cdf(-abs(z))

print("z-statistic:", z)
print("p-value:", p_value)
```

```python
x = np.arcsinh(df["area"])
y = np.arcsinh(df["price"])

data = pd.DataFrame({"x": x, "y": y}).dropna()

x = data["x"].to_numpy()
y = data["y"].to_numpy()
n = len(data)
```

```python
r_hat = np.corrcoef(x, y)[0, 1]
print("Sample correlation:", r_hat)
```

```python
R = 2000
results = []

for t in range(R):
    indices = rng.choice(n, size=n, replace=True)

    x_boot = x[indices]
    y_boot = y[indices]

    r_boot = np.corrcoef(x_boot, y_boot)[0, 1]
    results.append(r_boot)

results = np.array(results)
se = results.std(ddof=1)

print("Bootstrap SE:", se)
```

```python
for z_star in [1.645, 1.96, 2.576]:
    lower = r_hat - z_star * se
    upper = r_hat + z_star * se
    print(f"CI: ({lower:.4f}, {upper:.4f})")
```

```python
z = (r_hat - 0) / se
p_value = 2 * norm.cdf(-abs(z))

print("z-statistic:", z)
print("p-value:", p_value)
```

```python
df = pd.read_csv('./data/metabric.csv')

print(df['Overall Survival Status'].value_counts())
print(df["Overall Survival (Months)"].isna().sum())
```

```python
x_bar = df["Overall Survival (Months)"].mean()
x_bar
```

```python
rng = np.random.default_rng(42)
R = 2_000

x = df["Overall Survival (Months)"].to_numpy()


results = []
for t in range(R):
    bootstrap_sample = rng.choice(x, size=len(x), replace=True)
    x_bar_boot = bootstrap_sample.mean()
    results.append(x_bar_boot)
results = np.array(results)
```

```python
se = results.std(ddof=1)
se
```

```python
for z_star in [1.645, 1.96, 2.576]:
    lower = x_bar - z_star * se
    upper = x_bar + z_star * se
    print(f"CI ({z_star}): ({lower:.4f}, {upper:.4f})")
```

```python
z = (x_bar - 120) / se
p_value = 2 * norm.cdf(-abs(z))

print("z-statistic:", z)
print("p-value:", p_value)
```
